```python

print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

if duplicate_count == 0:
    print("No duplicate records found.")
else:
    df = df.drop_duplicates()
    print("Duplicate records removed.")
```

```python

required_columns = [
    "product_type",
    "product_name",
    "brand",
    "marketplace",
    "price"
]

print("Missing values in required columns:")
print(df[required_columns].isnull().sum())

# Remove rows without essential product information
df = df.dropna(subset=required_columns)

# Convert price and rating to numeric values
df["price"] = pd.to_numeric(df["price"], errors="coerce")
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

# Remove rows with invalid prices
df = df.dropna(subset=["price"])
df = df[df["price"] > 0]

print("\nData cleaning completed!")
print("Remaining records:", len(df))
```

```python

print("Cleaned Dataset:")
display(df)

print("\nFinal Dataset Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)

df.to_csv("mobile_laptop_prices_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```

```python

# Average price for each product category

category_prices = df.groupby("product_type")["price"].mean()

print("Average Price by Product Category:")
display(category_prices.round(2))
```

```python

brand_prices = (
    df.groupby(["product_type", "brand"])["price"]
    .agg(["mean", "min", "max", "count"])
    .round(2)
)

brand_prices = brand_prices.rename(columns={
    "mean": "Average Price",
    "min": "Lowest Price",
    "max": "Highest Price",
    "count": "Number of Listings"
})

print("Brand Price Analysis:")
display(brand_prices)
```

```python

comparison = df.pivot_table(
    index=["product_type", "product_name", "brand", "ram", "storage"],
    columns="marketplace",
    values="price",
    aggfunc="mean"
).reset_index()

if "Amazon" in comparison.columns and "Flipkart" in comparison.columns:
    comparison["Price Difference"] = (
        comparison["Amazon"] - comparison["Flipkart"]
    )

    comparison["Cheaper Marketplace"] = np.select(
        [
            comparison["Amazon"] < comparison["Flipkart"],
            comparison["Amazon"] > comparison["Flipkart"]
        ],
        ["Amazon", "Flipkart"],
        default="Same Price"
    )

print("Marketplace Price Comparison:")
display(comparison)
```

```python

for category in ["Mobile", "Laptop"]:
    category_df = df[df["product_type"] == category]

    print(f"\nCheapest {category} Listings:")

    if not category_df.empty:
        display(
            category_df.nsmallest(3, "price")[
                ["product_name", "brand", "marketplace", "price"]
            ]
        )
    else:
        print("No products found.")
```

```python

comparison.to_csv("marketplace_price_comparison.csv", index=False)

brand_prices.to_csv("brand_price_analysis.csv")

print("Analysis files saved successfully!")
```

```python

%%writefile app.py
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

st.set_page_config(
    page_title="Mobile & Laptop Price Analysis",
    page_icon="💻",
    layout="wide"
)

st.title("📱 Mobile & Laptop Price Comparison Dashboard")
st.write("Explore product prices, brands, and marketplace differences.")

# Load dataset
@st.cache_data
def load_data():
    return pd.read_csv("mobile_laptop_prices_cleaned.csv")

df = load_data()

# Sidebar filters
st.sidebar.header("Filter Products")

product_types = st.sidebar.multiselect(
    "Product Type",
    options=sorted(df["product_type"].dropna().unique()),
    default=sorted(df["product_type"].dropna().unique())
)

brands = st.sidebar.multiselect(
    "Brand",
    options=sorted(df["brand"].dropna().unique()),
    default=sorted(df["brand"].dropna().unique())
)

marketplaces = st.sidebar.multiselect(
    "Marketplace",
    options=sorted(df["marketplace"].dropna().unique()),
    default=sorted(df["marketplace"].dropna().unique())
)

filtered_df = df[
    df["product_type"].isin(product_types)
    & df["brand"].isin(brands)
    & df["marketplace"].isin(marketplaces)
]

# Summary metrics
col1, col2, col3 = st.columns(3)

col1.metric("Total Listings", len(filtered_df))

if not filtered_df.empty:
    col2.metric(
        "Average Price",
        f"₹{filtered_df['price'].mean():,.0f}"
    )
    col3.metric(
        "Lowest Price",
        f"₹{filtered_df['price'].min():,.0f}"
    )
else:
    col2.metric("Average Price", "—")
    col3.metric("Lowest Price", "—")

st.divider()

# Product table
st.subheader("Product Listings")
st.dataframe(filtered_df, use_container_width=True)

# Price chart
if not filtered_df.empty:
    st.subheader("Price Comparison by Product")

    chart_data = filtered_df.copy()
    chart_data["Product Label"] = (
        chart_data["product_name"]
        + " ("
        + chart_data["marketplace"]
        + ")"
    )

    chart_data = chart_data.sort_values("price")

    fig, ax = plt.subplots(figsize=(10, 5))
    sns.barplot(
        data=chart_data,
        x="price",
        y="Product Label",
        ax=ax
    )
    ax.set_xlabel("Price (₹)")
    ax.set_ylabel("Product")
    ax.set_title("Listed Prices")
    fig.tight_layout()

    st.pyplot(fig)

    st.subheader("Average Price by Brand")
    brand_avg = (
        filtered_df.groupby("brand", as_index=False)["price"]
        .mean()
        .sort_values("price")
    )

    st.bar_chart(brand_avg, x="brand", y="price")

    st.subheader("Marketplace Summary")
    marketplace_avg = (
        filtered_df.groupby("marketplace", as_index=False)["price"]
        .mean()
    )

    st.bar_chart(marketplace_avg, x="marketplace", y="price")
else:
    st.warning("No products match your selected filters.")

st.caption(
    "Prices are based on the dataset provided. "
    "Verify current listings before making purchasing decisions."
)
```
